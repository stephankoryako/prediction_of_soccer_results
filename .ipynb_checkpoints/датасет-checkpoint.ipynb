## Рейтинг Эло

```python
start_elo = ids
lig = []
for i in lst:
    l = data.loc[data['HomeTeam'] == i]['league'].unique()[0]
    lig.append(l)
lig
def f(column):
    if column == 'Premier league':
        column = 2600
    elif column == 'LaLiga':
        column = 2550
    elif column == 'Seria A':
        column = 2500
    elif column == 'Bundesliga':
        column = 2450
    elif column == 'League 1':
        column = 2400
    elif column == 'Championship':
        column = 2300
    elif column == 'Segunda':
        column = 2250
    elif column == 'Seria B':
        column = 2200
    elif column == '2.Bundesliga':
        column = 2150
    elif column == 'League 2':
        column = 2100
    elif column == 'NaN':
        column = 'NaN'
    return column
pd.set_option('mode.chained_assignment',None)
start_elo['start_rating'] = [f(x) for x in lig]
start_elo.columns = ['team', 'id', 'start_rating']
```

```python
start_elo.to_csv('start_ratings.csv', index=False)
```

```python
second_stat
```

```python
match_stat = pd.concat([first_stat, second_stat], ignore_index = True)
```

```python
match_stat = match_stat[['match_id','team','total_shots', 'shots_on_goal', 'fouls', 'corners', 'yellow_cards']]
```
